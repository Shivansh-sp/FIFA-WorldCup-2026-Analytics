```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("../data/fifa_world_cup_2026_cleaned.csv")
```

```python
df["match_date"] = pd.to_datetime(df["match_date"])
```

```python
df["goal_conversion_rate"] = np.where(
    df["shots"] > 0,
    df["goals"] / df["shots"],
    0
)
```

```python
df["shot_accuracy"] = np.where(
    df["shots"] > 0,
    df["shots_on_target"] / df["shots"],
    0
)
```

```python
df["pass_completion_rate"] = np.where(
    df["total_passes"] > 0,
    df["successful_passes"] / df["total_passes"],
    0
)
```

```python
df["dribble_success_rate"] = np.where(
    df["dribbles_attempted"] > 0,
    df["successful_dribbles"] / df["dribbles_attempted"],
    0
)
```

```python
df["cross_accuracy"] = np.where(
    df["crosses"] > 0,
    df["successful_crosses"] / df["crosses"],
    0
)
```

```python
df["aerial_duel_win_rate"] = np.where(
    (df["aerial_duels_won"] + df["aerial_duels_lost"]) > 0,
    df["aerial_duels_won"] /
    (df["aerial_duels_won"] + df["aerial_duels_lost"]),
    0
)
```

```python
df["goal_contribution"] = df["goals"] + df["assists"]
```

```python
df["offensive_actions"] = (
    df["goals"] +
    df["assists"] +
    df["key_passes"] +
    df["shots"]
)
```

```python
df["total_defensive_actions"] = (
    df["tackles"] +
    df["interceptions"] +
    df["clearances"] +
    df["blocks"]
)
```

```python
df["physical_work_rate"] = (
    df["distance_covered_km"] +
    df["sprint_distance_km"]
)
```

```python
df["discipline_score"] = (
    df["yellow_cards"] +
    (2 * df["red_cards"])
)
```

```python
df["minutes_per_goal"] = np.where(
    df["goals"] > 0,
    df["minutes_played"] / df["goals"],
    df["minutes_played"]
)
```

```python
new_features = [
    "goal_conversion_rate",
    "shot_accuracy",
    "pass_completion_rate",
    "dribble_success_rate",
    "cross_accuracy",
    "aerial_duel_win_rate",
    "goal_contribution",
    "offensive_actions",
    "total_defensive_actions",
    "physical_work_rate",
    "discipline_score",
    "minutes_per_goal"
]

df[new_features].head()
```

```python
df.to_csv("../data/fifa_world_cup_2026_feature_engineered.csv", index=False)
```
